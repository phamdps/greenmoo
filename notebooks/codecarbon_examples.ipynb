```python
import time
import numpy as np
from codecarbon import EmissionsTracker, track_emissions

# ==============================================================================
# Method 1: Explicit Tracking with EmissionsTracker (Best for Scripts/Pipelines)
# ==============================================================================
print("--- Method 1: Using EmissionsTracker Class ---")

# 1. Initialize the tracker
tracker = EmissionsTracker(
    project_name="matrix_multiplication_benchmark",
    output_dir="../data/results/emissions_report_1",  # Where to save the emissions.csv report
    measure_power_secs=1                             # Read hardware sensors every 1 second
)

# 2. Start monitoring
tracker.start()

try:
    print("Running heavy computation...")
    # Simulating a heavy CPU task (Large Matrix Operations)
    matrix_size = 4000
    A = np.random.rand(matrix_size, matrix_size)
    B = np.random.rand(matrix_size, matrix_size)
    C = np.dot(A, B)
    
finally:
    # 3. Stop monitoring and output results
    emissions: float = tracker.stop()
    print(f"\n[Result] Total CO2 Emissions: {emissions:.6f} kg CO2eq")
```

```python
# ==============================================================================
# Method 2: Function Tracking with Decorator (Best for Modular Code)
# ==============================================================================
print("\n--- Method 2: Using @track_emissions Decorator ---")

@track_emissions(project_name="light_task_function", output_dir="../data/results/emissions_report_2")
def process_data():
    print("Running light task...")
    time.sleep(3) # Simulating waiting/light processing

# Just call the function as usual
process_data()
```
